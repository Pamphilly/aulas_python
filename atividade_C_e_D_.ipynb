```python
pares = 0

for i in range(1, 6):
    num = int(input(f"Digite o {i} número: "))
    
    if num % 2 == 0:
        pares += 1

print(f"Você digitou {pares} números pares.")
```

```python
nota = float(input("Digite uma nota (entre 0 e 10): "))

while nota < 0 or nota > 10:
    print("Nota inválida!")
    nota = float(input("Tente novamente!: "))

print(f"Nota", {nota}, "registrada com sucesso!")
```

```python
idade = int(input("Digite sua idade: "))
tem_documento = input("Você está com o documento? (sim/nao): ").strip().lower() #eu pesquisei, e essas funções evita erro caso o usuario digite "sim" com espaços ou maiúsculo

if idade >= 18 and tem_documento == "sim":
    print("Acesso autorizado! Bem-vindo.")
else:
    print("Acesso negado. É necessário ter 18 anos e apresentar o documento.")
```

```python
usuario_correto = "Ujeferson"
senha_correta = "1234"

tentativas = 3

while tentativas > 0:
    print(f"\nTentativas restantes: {tentativas}")
    user = input("Usuário: ")
    senha = input("Senha: ")

    if user == usuario_correto and senha == senha_correta:
        print("Login realizado com sucesso! Bem-vindo.")
        break  
    else:
        tentativas -= 1
        print("Usuário ou senha incorretos.")

if tentativas == 0:
    print("\nAcesso bloqueado! Você excedeu o limite de 3 tentativas.")
```
