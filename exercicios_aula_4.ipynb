```python
numero = int(input("Digite um numero: "))

if numero >= 50 and numero <=100:
    print("Numero esta entre 50 e 100")

else:
    print("Nao esta entre 50 e 100")
```

```python
dia_semana = input (" Digite o dia da semana : ")

if dia_semana == "sabado" or dia_semana == "domingo":
    print ("É final de semana ")
else :
    print ("É dia útil ")
```

```python
usuario = input("Digite o usuario: ")
senha = input("Digite a senha: ")

if usuario == "Ryan" and senha == "123":
    print("Loguin realizado")

else:
    print("Usuario ou senha invalidos!")
```

```python
nota = int(input("Digite sua nota: "))
frequencia = int(input("Digite sua frequencia: "))

if nota >= 7 and frequencia >= 70:
    print("Aprovado!")

else:
    print("Reprovado!")
```

```python
aluno = input("Voce entregou o trabalho? (sim/nao) ")

if not aluno == 'sim':
    print("Voce nao entregou o trabalho!")

else:
    print("Voce entregou o trabalho")
```

```python
contador = 1

while contador <= 10:
    print(contador)
    contador += 1  
```

```python
senha = int(input("Digite uma senha numérica: "))

while senha < 1000 or senha > 1500:
    print("Senha incorreta, tente novamente!")
    senha = int(input("Digite uma senha numérica: "))

print("Senha correta!")
    
```

```python
num = int(input("Digite um numero: "))

while num != 0:
    print("Numero incorreto, tente novamente!")
    num = int(input("Digite outro numero: "))

print("Numero correto!")
```

```python
contador = 10

while contador >= 1:
    print(contador)
    contador -= 1  

print("Fim da contagem!")
```

```python
while True:
    idade = int(input("Digite sua idade: "))
    
    if idade > 0:
        break  
    else:
        print("Erro: A idade deve ser um número maior que zero.")

print("Idade correta!")
```

```python
from collections import deque
fila = deque()


while True:
    
    print("\n=====Menu=====")
    print("1- Cadastrar nome")
    print("0- Sair")

    opcao = input("Digite uma opcao: ")
    if opcao == "1":
    
        nome = input("Digite seu nome")
        fila.append(nome)
        print("Nome cadastrado")

    elif opcao == "0":
        print("Encerrando Menu")
        break
    else:
        print("Opcao invalida!")
```
